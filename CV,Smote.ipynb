Cross Validation
            -Is the technique to check how the model is well performed for unseen data
            -Dataset SPlit
                    Train 80% and Test 20%
                    Model is trained for train data
                    Evaluate for test data once
                    Result may varying depending on the split
            -K Fold Cross Validation
                  Dataset (1000 records)
                  Split the Data into 5 K Folds
                  K=1      k=2.......K=5
                  200      200       200
            -5 Fold Cross Validation Process 
                -Iteration 1
                    F1,F2,F3,F4,F5
                    Validation Set-F1
                    Train Data-F2,F3,F4,F5
                -Iteration 2
                    F1,F2,F3,F4,F5
                    Validation Set-F2
                    Train Data-F1,F3,F4,F5
                -Iteration 3
                    F1,F2,F3,F4,F5
                    Validation Set-F3
                    Train Data-F1,F2,F4,F5
                -Iteration 4
                    F1,F2,F3,F4,F5
                    Validation Set-F4
                    Train Data-F2,F3,F1,F5
                -Iteration 5
                    F1,F2,F3,F4,F5
                    Validation Set-F5
                    Train Data-F2,F3,F4,F1
                Calculate the performance for all the folds and calculate the average
                Final Cross Validation Score:Average of 5 Score

                Example
                Fold       Accuracy   Precision Recall F1 Score
                1           0.90        0.91     0.90   0.91
                2           0.90        0.90     0.91   0.90
                ..
                5
                Avg         0.90        0.91       0.90  0.91

SVM
    -Support Vector Machine
    -Supervised Learning
        -Classification-SVC
        -Regression-SVR
    -SVM finds the best boundary(Hyperplane) that separates different classes of data with maximum marigin
    -Support
            -The important data points that help decision boundary
    -Vector
            -Each data can be represented as a vectory values
    -Machine
            -Algorithm that finds the optimal decision boundary
    -Always take Maximum Margin
            -SVM chooses the best decision boundary for large separation class
    -Soft Margin Vs Hard Margin
        Hard Margin
            -No training data is allowed to be on the wrong side
            -It works only when ur data is perfectly separable
        Soft Margin
            -It allows the misclassification
    -Hyperparameter
            -Before evaluating the model we will figure out which parameter gives the best result
            -C=Regularization Parameter
                Control
                    -Large Margin
                    -Training Errors
                Small C
                    -Allows more misclassification
                Large C
                    -Penalize misclassification heavily
    -SVM CHoose Kernel
        -Straight line cannot handle data perfectly
        -SVM uses Kernel Function to handle non-linear relationship
    -Types of Kernel
        -Linear Kernel-Using a straight line
        -Polynomial Kernel-Curved Decision Boundary
        -RBF(Radial Basis Function)-Non-Linear Classification
        -Sigmoid-Sigmoid Function common to RBF
    -Gamma Value Hyperparameters
        Kernel-RBF,Sigmoid,Polynomial
        Gamma-how closely does the model focus on individual data points??
        Low Gamma->Large Influence area-smoother decision boundary
        High Gamma->Small Influence area-complex decision boundary
