PCA
    -Principal Component Analysis
    -Unsupervised Learning
    -Reduce the features based on the existing data
    -Dimensionality Reduction Technique
    -PCA reduces the number of features in a dataset while keeping as much important information as possible.
    -20 features -4 features
    -PC1,Pc2,PC3 and PC4 -should have high variance information of data
    -Variance
            -Covariance-Find the relationship b/w among the variable
            -How much data is spread out
    -PCA it takes the data which consists of high variance information
    -Explained Variance Ration
        -How much information/variance in each principal component
        PC1-50%
        PC2-25%
        PC3-30%
        PC4-25%
    -Cummulative Sum
        PC1+PC3+PC4=100% variance
    PCA Steps
        1.Collect the data
        2.Handle Missing values
        3.Standarize the data
        4.Calculate Covariance
        5.Calculate Eigenvalue/Eigenvector-to show in which direction PCA working either its highest or lowest
        6.Sort the Eigenvalues-Descending to ascending PC1-90%,70,60,55
        7.Select the component-cumsum
        8.Reduce the features

```python
X=data.drop("Class",axis=1)
y=data["Class"]
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
```

```python
scaled_data=scaler.fit_transform(X)
```

```python
## getting the optimal number of pca
from sklearn.decomposition import PCA
pca = PCA()
principalComponents = pca.fit_transform(X)
```

```python
pdf1=pd.DataFrame(data=principalComponents)
```

```python
import matplotlib.pyplot as plt
plt.figure()
plt.plot(np.cumsum(pca.explained_variance_ratio_))

plt.xlabel('Number of Components')
plt.ylabel('Variance (%)') #for each component
plt.title('Explained Variance')
plt.show()
```

```python
pca = PCA(n_components= 4)
new_data = pca.fit_transform(X)
```

```python
principal_Df = pd.DataFrame(data = new_data
             , columns = ['PC1', 'PC2','PC3','PC4'])
```
